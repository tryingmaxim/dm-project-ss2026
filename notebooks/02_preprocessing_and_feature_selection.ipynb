# Notebook 02 â€“ Preprocessing & Feature Selection

**Research question:** Which demographic, socioeconomic, and academic factors most strongly predict student dropout?

This notebook covers:
1. Binary target preparation (Dropout vs. Graduate)
2. Missing-value analysis
3. Descriptive statistics
4. Feature importance ranking via Mutual Information, ANOVA F-score, and Random Forest
5. Feature group comparison
6. Selection of Top-10 and Top-15 feature subsets for downstream modelling

```python
import sys, os
# Make sure the project root is on the path
sys.path.insert(0, os.path.abspath('..'))

import pandas as pd
import matplotlib
matplotlib.use('Agg')
import matplotlib.pyplot as plt

from src.data_loader import load_raw, describe_dataset, prepare_binary, get_X_y
from src.preprocessing import check_missing
from src.feature_selection import (
    compute_all_importances, get_top_features, group_summary
)
from src.evaluate_models import descriptive_statistics
from src.plots import (
    plot_class_balance_binary,
    plot_feature_importance_top,
    plot_group_importance,
)
from src.config import TOP_N_FEATURES

pd.set_option('display.max_columns', 50)
pd.set_option('display.float_format', '{:.4f}'.format)
```

## 1. Load & Inspect Raw Data

```python
df_raw = load_raw()
describe_dataset(df_raw)
df_raw.head(3)
```

## 2. Binary Target Preparation

Remove *Enrolled* records; encode **Dropout = 1**, **Graduate = 0**.

```python
df = prepare_binary(df_raw)
print(f'Binary dataset shape: {df.shape}')
print(df['Target'].value_counts().rename({1: 'Dropout', 0: 'Graduate'}))
```

```python
X, y = get_X_y(df)
print(f'Features: {X.shape[1]}  |  Samples: {X.shape[0]}')

fig = plot_class_balance_binary(y)
plt.show()
```

## 3. Missing Values & Descriptive Statistics

```python
missing = check_missing(X)
if missing.empty:
    print('No missing values detected.')
else:
    print(missing)
```

```python
stats = descriptive_statistics(df, save=True)
stats
```

## 4. Feature Importance

Three complementary methods:
- **Mutual Information** (information gain, model-agnostic)
- **ANOVA F-score** (linear relationship, fast)
- **Random Forest Importance** (non-linear, embedded)

```python
importances = compute_all_importances(X, y, save=True)
```

```python
print('=== Top 15 â€“ Mutual Information ===')
print(importances['mutual_info'].head(15).to_string(index=False))
```

```python
print('=== Top 15 â€“ ANOVA F-score ===')
print(importances['fscore'].head(15).to_string(index=False))
```

```python
print('=== Top 15 â€“ Random Forest Importance ===')
print(importances['rf_importance'].head(15).to_string(index=False))
```

```python
# Visualise top-10 by Mutual Information
fig = plot_feature_importance_top(
    importances['mutual_info'], 'mi_score',
    title='Top-10 Features â€“ Mutual Information',
    n=10, fname='feature_importance_top10_mi.png'
)
plt.show()

# Visualise top-10 by Random Forest
fig = plot_feature_importance_top(
    importances['rf_importance'], 'rf_importance',
    title='Top-10 Features â€“ Random Forest Importance',
    n=10, fname='feature_importance_top10.png'
)
plt.show()
```

## 5. Feature Group Comparison

Tests the hypothesis: *academic > socioeconomic > demographic*.

```python
grp_summary = group_summary(importances['mutual_info'], score_col='mi_score', save=True)
print(grp_summary.to_string(index=False))

fig = plot_group_importance(grp_summary, score_col='mean_score')
plt.show()
```

## 6. Top-N Feature Subsets

The top-10 and top-15 feature lists (by Mutual Information) are persisted
for use in Notebook 03.

```python
import json, os
from src.config import TABLES_DIR

top_features = {}
for n in TOP_N_FEATURES:
    feats = get_top_features(importances['mutual_info'], 'mi_score', n)
    top_features[f'top{n}'] = feats
    print(f'Top-{n} features (MI): {feats}')

os.makedirs('../results/tables', exist_ok=True)
with open('../results/tables/top_feature_subsets.json', 'w') as f:
    json.dump(top_features, f, indent=2)

print('\nTop feature subsets saved.')
```
