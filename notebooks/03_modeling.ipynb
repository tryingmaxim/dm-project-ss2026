# Notebook 03 - Model Training and Cross-Validation
**Research question:** Which classification algorithm achieves the best predictive performance under 10-fold cross-validation?
This notebook:
1. Loads the binary dataset.
2. Runs 10-fold cross-validation on all features.
3. Repeats the experiment on the Top-10 feature subset.
4. Visualizes stability across folds.
5. Compares all features vs. Top-10 features.
6. Identifies and persists the best model.

```python
import sys, os
sys.path.insert(0, os.path.abspath('..'))
import json
import pandas as pd
import matplotlib
matplotlib.use('Agg')
import matplotlib.pyplot as plt
from src.data_loader import load_raw, prepare_binary, get_X_y
from src.evaluate_models import best_model_name, fit_and_save_best
from src.train_models import run_cv
from src.plots import (
    plot_cv_boxplot,
    plot_feature_subset_comparison,
    plot_model_comparison,
)
pd.set_option('display.float_format', '{:.4f}'.format)
```

## 1. Load Binary Dataset

```python
df = prepare_binary(load_raw())
X_all, y = get_X_y(df)
print(f'Samples: {X_all.shape[0]}  |  Features: {X_all.shape[1]}')
print(f'Class balance - Dropout: {y.sum()}  |  Graduate: {(y == 0).sum()}')
```

## 2. 10-Fold CV - All Features

```python
print('Running 10-fold CV on all features...\n')
results_all = run_cv(X_all, y, label='all_features', save=True)
results_all[['model', 'accuracy_mean', 'accuracy_std', 'f1_mean', 'f1_std', 'roc_auc_mean']]
```

```python
fig = plot_model_comparison(results_all, metric='accuracy')
plt.show()
fig = plot_model_comparison(results_all, metric='f1')
plt.show()
```

## 3. Cross-Validation Stability

```python
with open('../results/tables/cv_fold_scores.json', encoding='utf-8') as f:
    cv_fold_scores = json.load(f)
fig = plot_cv_boxplot(cv_fold_scores, metric='f1', fname='cv_boxplot_f1.png')
plt.show()
fig = plot_cv_boxplot(cv_fold_scores, metric='roc_auc', fname='cv_boxplot_roc.png')
plt.show()
```

## 4. 10-Fold CV - Top-10 Features
Load the Top-10 feature subset created in Notebook 02 and compare performance.

```python
with open('../results/tables/top_feature_subsets.json', encoding='utf-8') as f:
    top_features = json.load(f)
top10 = top_features['top10']
print('Top-10 features:', top10)
X_top10 = X_all[top10]
print('\nRunning 10-fold CV on Top-10 features...\n')
results_top10 = run_cv(X_top10, y, label='top10', save=True)
results_top10[['model', 'accuracy_mean', 'accuracy_std', 'f1_mean', 'f1_std', 'roc_auc_mean']]
```

```python
fig = plot_feature_subset_comparison(results_all, results_top10, metric='f1_mean')
plt.show()
```

## 5. Best Model

```python
best = best_model_name(results_all, metric='f1_mean')
print(f'Best model (by F1): {best}')
best_row = results_all[results_all['model'] == best].iloc[0]
print(f"  Accuracy : {best_row['accuracy_mean']:.4f} +/- {best_row['accuracy_std']:.4f}")
print(f"  F1       : {best_row['f1_mean']:.4f} +/- {best_row['f1_std']:.4f}")
print(f"  ROC-AUC  : {best_row['roc_auc_mean']:.4f} +/- {best_row['roc_auc_std']:.4f}")
fitted_model = fit_and_save_best(X_all, y, best)
print('Best model saved to results/models/best_model.pkl')
```
